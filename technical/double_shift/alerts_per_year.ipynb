```python
import matplotlib.pylab as plt 
%matplotlib inline
import numpy as np
import rubin_sim.splat as splat
import pandas as pd
import sqlite3
import healpy as hp
import copy
import datetime
import os

from rubin_sim.splat.plots import bright_filter_colors

from rubin_scheduler.utils import (
    angular_separation,
    ddf_locations,hpid2_ra_dec)

from os.path import basename

```

```python
def len_df(df_in):
    return len(df_in.index)
```

```python
db_files = ["double_shift_v5.3.7_10yrs.db", "baseline_v5.3.7_10yrs.db"]
results = []
for db_file in db_files:
    data_source = basename(db_file).replace('.db', '')
    con = sqlite3.connect(db_file)
    df = pd.read_sql('select * from observations;', con)
    con.close()
    # Convert to a numpy array
    # visits_array = df.to_records(index=False)
    df["year"] = np.floor(df["night"].values/365.25).astype(int)
    
    # reasonable Keck limit I think. Hawaii is ~20 deg North
    # US SW is around 30 deg north
    for dec_limit in [-30]:
        indx = np.where((df["fieldDec"] > dec_limit) & (df["visitExposureTime"] > 20.))[0]
        result = df.iloc[indx].groupby("year").apply(len_df)
        
        #result["dec_limit"] = dec_limit
        #result["run_name"] = data_source
        results.append(result)
```
