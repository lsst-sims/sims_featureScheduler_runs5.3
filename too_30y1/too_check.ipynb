```python
import pandas as pd
import numpy as np
import sqlite3
```

```python

for dbfile in ["too_30sy1_v5.3.6_10yrs.db", "baseline_v5.3.6_10yrs.db"]:

    print("-------")
    con = sqlite3.connect(dbfile)
    df = pd.read_sql("select * from observations where night < 730;", con)
    con.close()
    
    
    too_obs = ["ToO" in note for note in df["scheduler_note"]]
    print("total ToO exposure time", np.sum(df["visitExposureTime"].values[too_obs]))
    print("N visits", np.sum(too_obs))
    
    y1 = np.where((too_obs) & (df["night"].values < 366))[0]
    print("total ToO exposure time, y1", np.sum(df["visitExposureTime"].values[y1]))

    print("total ToO exp time", np.sum(df["visitExposureTime"].values[too_obs])/3600/24)
    print("total exp time", np.sum(df["visitExposureTime"].values))
    
    ack = []
    for ra, band in zip(df["fieldRA"].values, df["band"].values):
        ack.append(str(ra) + ',' + band)
    
    print("n repeat band+ra=",  np.size(ack)-np.size(np.unique(ack)))
```
