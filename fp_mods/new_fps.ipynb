```python
import healpy as hp
%matplotlib inline
from new_footprints import FootprintMod1, FootprintMod2, FootprintMod3, FootprintMod4, FootprintMod5
import numpy as np
from rubin_scheduler.scheduler.utils import Phase3AreaMap
```

```python
sag = Phase3AreaMap()
sky_maps, labels = sag.return_maps()

label_int = np.zeros(hp.nside2npix(32))
for i,label in enumerate(np.unique(labels)):
    indx = np.where(labels == label)[0]
    label_int[indx] = i
hp.mollview(label_int, title="Baseline")
print("area in bulgy Galactic Plane", np.size(np.where(labels == "bulgy")[0]) * hp.nside2pixarea(sag.nside, degrees=True))
```

```python
sag = FootprintMod1()
sky_maps, labels = sag.return_maps()

label_int = np.zeros(hp.nside2npix(32))
for i,label in enumerate(np.unique(labels)):
    indx = np.where(labels == label)[0]
    label_int[indx] = i
    print(i, label)
hp.mollview(label_int, title="mod 1: Like Baseline, dust limit down to 0.11")
print("area in bulgy Galactic Plane", np.size(np.where(labels == "bulgy")[0]) * hp.nside2pixarea(sag.nside, degrees=True))
```
