```python
import healpy as hp
%matplotlib inline
from new_footprints import FootprintMod1, FootprintMod2
import numpy as np
```

```python
sag = FootprintMod1()
sky_maps, labels = sag.return_maps()

label_int = np.zeros(hp.nside2npix(32))
for i,label in enumerate(np.unique(labels)):
    indx = np.where(labels == label)[0]
    label_int[indx] = i
    print(i, label)
hp.mollview(label_int, title="mod 1")
```
