```python
import numpy as np
import sqlite3
import pandas as pd
```

```python
con = sqlite3.connect("baseline_v5.3.3_10yrs.db")
df = pd.read_sql("select * from observations;", con)
con.close()
```

```python
note_root = np.array([note.split(",")[0].split(" ")[0].split(":")[0] for note in df["scheduler_note"]])
```

```python
np.unique(note_root)
```

```python
size = np.sum(df["visitExposureTime"])
for note in np.unique(note_root):
    indx = np.where(note_root == note)[0]
    print(note, np.sum(df["visitExposureTime"].values[indx])/size *100)
```
